# UNSW-NB15 Dataset Analysis

Load the training split and inspect its dimensions, schema, missing values, and binary target balance. Place the official CSV in `../dataset/` before running the analysis.

```python
from pathlib import Path
import pandas as pd

project_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
training_path = project_root / 'dataset' / 'UNSW_NB15_training-set.csv'
if not training_path.exists():
    raise FileNotFoundError(f'Missing dataset file: {training_path}')

data = pd.read_csv(training_path)
data.head()
```

```python
print(f'Rows: {len(data):,}')
print(f'Columns: {len(data.columns):,}')
display(data.dtypes.value_counts().rename_axis('dtype').to_frame('column_count'))
display(data.isna().sum().sort_values(ascending=False).head(15).to_frame('missing_values'))
```

```python
if 'label' not in data.columns:
    raise KeyError("Expected the binary target column 'label'.")
target_counts = data['label'].value_counts(dropna=False).sort_index()
display(target_counts.rename_axis('label').to_frame('count'))
display((target_counts / target_counts.sum()).rename('fraction').to_frame())
```
